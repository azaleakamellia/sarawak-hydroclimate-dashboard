Inject the in-situ data

```python
# ==============================================================================
# INJECT IN-SITU DATA AND STATION NAMES INTO MASTER PARQUET
# ==============================================================================
import sys
from pathlib import Path

# Set project root (since notebook is in /notebooks, root is one level up)
PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

# Import your updated injection function
from src.insitu_utils import inject_insitu_into_master

# Map the exact file paths based on your folder structure
MASTER_PARQUET = PROJECT_ROOT / "outputs" / "processed" / "sarawak_hydro_data_FULL.parquet"
INSITU_CSV     = PROJECT_ROOT / "data" / "gauges" / "cleaned_insitu_swk_1983_2024.csv"
GAUGE_META_CSV = PROJECT_ROOT / "data" / "gauges" / "sarawak_all_gauges.csv"
OUTPUT_PARQUET = PROJECT_ROOT / "outputs" / "processed" / "sarawak_MASTER_with_insitu.parquet"

# Execute the injection
df_final = inject_insitu_into_master(
    master_parquet_path=str(MASTER_PARQUET),
    insitu_csv_path=str(INSITU_CSV),
    gauge_metadata_path=str(GAUGE_META_CSV),
    output_parquet_path=str(OUTPUT_PARQUET)
)

# Display a quick preview of the results
print("\n--- Final Dataset Preview ---")
display(df_final.head())
```
